```r
matriz_aleatoria <- function(n, min = 1, max = 100) {
  matrix(
    sample(min:max, n^2, replace = TRUE),
    nrow = n,
    ncol = n
  )
}
matriz_aleatoria <- function(n, min = 1, max = 100) {
  matrix(
    sample(min:max, n^2, replace = TRUE),
    nrow = n,
    ncol = n
  )
}
matriz_aleatoria(50)
matriz_aleatoria(50)

multiplicar_matrices <- function(n, min = 1, max = 100) {

  A <- matriz_aleatoria(n, min, max)
  B <- matriz_aleatoria(n, min, max)

  C <- A %*% B

  list(
    matriz_A = A,
    matriz_B = B,
    producto = C
  )
}

resultado <- multiplicar_matrices(50)

resultado$matriz_A
resultado$matriz_B
resultado$producto
```

```r
multiplicar_matrices(50)
multiplicar_matrices(100)
multiplicar_matrices(150)
multiplicar_matrices(200)
multiplicar_matrices(250)
multiplicar_matrices(300)
multiplicar_matrices(400)
multiplicar_matrices(500)

tamanos <- c(50, 100, 150, 200, 250, 300, 400, 500)

#Medir cuánto tarda cada multiplicación.

tiempos <- numeric(length(tamanos))

for(i in seq_along(tamanos)){

  tiempos[i] <- system.time({
    multiplicar_matrices(tamanos[i])
  })["elapsed"]

}
```

```r
resultados <- data.frame(
  tamano = tamanos,
  tiempo = tiempos
)

resultados
```

```r
library(ggplot2)

ggplot(
  resultados,
  aes(x = tamano, y = tiempo)
) +
  geom_point(size = 8) +
  geom_line() +
  labs(
    title = "Tiempo de multiplicación de matrices",
    x = "Tamaño de la matriz (n × n)",
    y = "Tiempo de ejecución (segundos)"
  )
```
